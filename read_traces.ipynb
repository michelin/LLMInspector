```python
# Retreive Span 

import phoenix as px

span = px.Client().get_spans_dataframe(project_name='Multi-Agent System', filter_condition="name == 'Supervisor'")
span_output = span.loc[span['context.trace_id'] == 'f1e3ebbacf16080764549941f282eb36', 'attributes.output.value'].iloc[0]
span_output
```

```python
# Retreive Output Messages from Span

import json
data = json.loads(span_output)
messages = data["messages"]
messages
```

```python
# Format Structure of Output Messages

import re

def replace_space_after_symbols(text):
    # This regex looks for either ' or } or ] followed by a space
    return re.sub(r"([\'\}\"\]]) ", r"\1, ", text)
```

```python
import ast
import json

def split_key_value_pairs(s):
    pairs = []
    current = ''
    depth = 0
    in_string = False
    string_char = ''

    for i, char in enumerate(s):
        if char in ['"', "'"]:
            if not in_string:
                in_string = True
                string_char = char
            elif string_char == char and (i == 0 or s[i - 1] != '\\'):
                in_string = False
        elif not in_string:
            if char in '([{':
                depth += 1
            elif char in ')]}':
                depth -= 1
            elif char == ',' and depth == 0:
                pairs.append(current.strip())
                current = ''
                continue
        current += char
    if current:
        pairs.append(current.strip())
    return pairs

def safe_eval(value):
    try:
        return ast.literal_eval(value)
    except Exception:
        return value

def string_to_dict(input_str):
    result = {}
    if input_str:
        pairs = split_key_value_pairs(input_str)
        for pair in pairs:
            if '=' in pair:
                key, value = pair.split('=', 1)
                key = key.strip()
                value = value.strip()
                parsed_value = safe_eval(value)
                if isinstance(parsed_value, str) and parsed_value.startswith('{') and parsed_value.endswith('}'):
                    parsed_value = string_to_dict(parsed_value[1:-1])
                result[key] = parsed_value
    return result

```

```python
messages1 = []
for message in messages:
    # Replace spaces after special characters
    messages1.append(replace_space_after_symbols(message))

messages1
```

```python
messages2 = [] 
for message in messages1:
    messages2.append(string_to_dict(message))

messages2
```

```python
# Convert messages into Trace messages

output = []
from langchain_core.messages import HumanMessage, AIMessage, ToolMessage

output.append(HumanMessage(**messages2[0]))
for message in messages2[1:]:
    if "tool_call_id" in message:
        output.append(ToolMessage(**message))
    else:
        if not message["additional_kwargs"]:
            message["additional_kwargs"] = {'tool_calls': [{'id': message["tool_calls"][0]["id"], 'function': {'arguments': message["tool_calls"][0]["args"], 'name': message["tool_calls"][0]["name"]}, 'type': 'function'}], 'refusal': None}

        output.append(AIMessage(**message))
```

```python
import os
import pickle

output_path = f'./output/traces-{span["context.trace_id"].iloc[0]}.pkl'


output_dir = os.path.dirname(output_path)


try:
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    
    with open(output_path, 'wb') as f:
        pickle.dump(output, f)
    print(f"File successfully saved to {output_path}")

except FileNotFoundError as fnf_error:
    print(f"FileNotFoundError: {fnf_error}")
except PermissionError as perm_error:
    print(f"PermissionError: {perm_error}")
except Exception as e:
    print(f"An error occurred: {e}")
```
